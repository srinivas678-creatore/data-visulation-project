```python

df1['Year'] = pd.to_datetime(df1['Total Waste (Tons)'])

df1 = df1.sort_values('Year')

plt.figure(figsize=(10,5))

plt.plot(df1['Year'], df1['Total Waste (Tons)'], linewidth=2)

plt.xlabel("Date")
plt.ylabel("Total Waste")
plt.title("Total Waste Trend Over Time")

plt.xticks(rotation=45)
plt.tight_layout()
plt.grid(True)
```
